```python
import pandas as pd
import numpy as np
```

```python
# Load the dataset
df = pd.read_csv('Titanic-Dataset.csv')

# First look at the data
df.head()
```

```python
# Check shape and basic info
print("Shape:", df.shape)
print("\nColumn Data Types:")
print(df.dtypes)
print("\nNull Values per Column:")
print(df.isnull().sum())
```

```python
# Descriptive statistics
df.describe()
```

```python

df['Age'].fillna(df['Age'].median(), inplace=True)

df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

df.drop(columns=['Cabin'], inplace=True)

print("Null Values After Cleaning:")
print(df.isnull().sum())
```

```python
print("Duplicates before:", df.duplicated().sum())
df.drop_duplicates(inplace=True)
print("Duplicates after:", df.duplicated().sum())
```

```python

df['Sex'] = df['Sex'].str.lower().str.strip()
df['Embarked'] = df['Embarked'].str.upper().str.strip()

print("Sex values:", df['Sex'].unique())
print("Embarked values:", df['Embarked'].unique())
```

```python
df['PassengerId'] = df['PassengerId'].astype(str)

print("Updated Data Types:")
print(df.dtypes)
```

```python

for col in ['Age', 'Fare']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"{col}: {len(outliers)} outliers found (lower={lower:.2f}, upper={upper:.2f})")
```

```python

for col in ['Age', 'Fare']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df[col] = df[col].clip(lower=lower, upper=upper)

print("Outliers capped successfully!")
print("\nNew Age range:", df['Age'].min(), "to", df['Age'].max())
print("New Fare range:", df['Fare'].min(), "to", df['Fare'].max())
```

```python

summary = {
    'Metric': ['Total Rows', 'Null Values', 'Duplicate Rows', 'Columns'],
    'Before': [891, 866, 0, 12],
    'After': [len(df), df.isnull().sum().sum(), df.duplicated().sum(), len(df.columns)]
}

summary_df = pd.DataFrame(summary)
print(summary_df)
```

```python

df.to_csv('Titanic-Cleaned.csv', index=False)
print("Cleaned dataset saved as Titanic-Cleaned.csv")
print("\nFinal Shape:", df.shape)
df.head()
```
